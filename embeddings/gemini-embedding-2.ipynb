```python
from google import genai
```

```python
client = genai.Client()
```

```python
example_text1 = "I have a cute cat named Mouchan."
example_text2 = "My pet kitty Mouchan is so cute."
example_texts = [example_text1, example_text2]
```

```python
result = client.models.embed_content(
    model="gemini-embedding-2",
    contents=example_texts
)
```

```python
print(result.embeddings)
```

```python
if result.embeddings == None:
    print("No embeddings found.")
else:
    values = [e.values for e in result.embeddings]
```
